```python
import numpy as np
import pandas as pd
from datetime import datetime, timedelta

np.random.seed(42)

# ===============================
# City center (Example: Cairo)
# ===============================
city_lat, city_lon = 30.0444, 31.2357

incident_types = ["Electricity", "Water", "Road", "Gas", "Waste"]
area_types = ["Residential", "Commercial"]

# ===============================
# Hour probability distribution
# ===============================
probs = np.array(
    [0.02]*6 +     # 00:00 - 05:00 (low activity)
    [0.05]*6 +     # 06:00 - 11:00 (morning)
    [0.06]*6 +     # 12:00 - 17:00 (peak)
    [0.04]*6       # 18:00 - 23:00 (evening)
)
probs = probs / probs.sum()  # normalize

data = []
start_date = datetime(2024, 1, 1)

# ===============================
# Generate Hotspot Records
# ===============================
for i in range(1, 60001):

    # --- Spatial clustering ---
    lat_offset = np.random.normal(0, 0.02)
    lon_offset = np.random.normal(0, 0.02)

    latitude = round(city_lat + lat_offset, 6)
    longitude = round(city_lon + lon_offset, 6)

    # --- Incident attributes ---
    incident_type = np.random.choice(incident_types)
    severity_score = round(np.random.beta(2, 2), 2)

    hour = np.random.choice(range(24), p=probs)
    day_of_week = np.random.randint(0, 7)
    month = np.random.randint(1, 13)

    area_type = np.random.choice(area_types, p=[0.7, 0.3])
    incident_count_nearby = np.random.poisson(lam=5)

    hour = int(np.random.choice(range(24), p=probs))

    timestamp = start_date + timedelta(
    days=int(np.random.randint(0, 365)),
    hours=hour)

    data.append({
        "incident_id": i,
        "latitude": latitude,
        "longitude": longitude,
        "incident_type": incident_type,
        "severity_score": severity_score,
        "hour": hour,
        "day_of_week": day_of_week,
        "month": month,
        "area_type": area_type,
        "incident_count_nearby": incident_count_nearby,
        "timestamp": timestamp
    })

# ===============================
# Save Dataset
# ===============================
df_hotspot = pd.DataFrame(data)
df_hotspot.to_csv("hotspot_dataset.csv", index=False)

print("✅ Hotspot dataset generated successfully!")
print("Shape:", df_hotspot.shape)
print(df_hotspot.head())
```
