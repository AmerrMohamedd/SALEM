```python
import numpy as np
import pandas as pd

np.random.seed(42)

severity_map = {"Low": 0.3, "Medium": 0.6, "High": 0.9}
teams = ["Team_A", "Team_B", "Team_C"]

data = []

for i in range(1, 12001):
    severity_label = np.random.choice(
        ["Low", "Medium", "High"], p=[0.4, 0.35, 0.25]
    )
    severity_score = severity_map[severity_label]

    area_load = round(np.random.beta(2, 2), 2)
    available_teams = np.random.randint(1, 5)

    team_skill_match = round(np.random.beta(5, 2), 2)
    historical_perf = round(np.random.beta(4, 2), 2)

    citizen_trust = round(np.random.uniform(0.3, 1.0), 2)

    # Priority Logic
    if severity_label == "High" and area_load > 0.6:
        priority = "P1"
    elif severity_label == "Low" and area_load < 0.4:
        priority = "P3"
    else:
        priority = "P2"

    assigned_team = np.random.choice(teams)
    resources = (
        "High" if priority == "P1"
        else "Medium" if priority == "P2"
        else "Low"
    )

    data.append({
        "incident_id": i,
        "severity_label": severity_label,
        "severity_score": severity_score,
        "area_load": area_load,
        "available_teams": available_teams,
        "team_skill_match": team_skill_match,
        "historical_team_performance": historical_perf,
        "citizen_trust_score": citizen_trust,
        "priority_level": priority,
        "assigned_team": assigned_team,
        "suggested_resources": resources
    })

df_priority = pd.DataFrame(data)
df_priority.to_csv("priority_allocation_dataset.csv", index=False)

print("✅ Priority dataset generated:", df_priority.shape)
print(df_priority.head())
```
