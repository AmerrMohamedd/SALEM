```python
import numpy as np
import pandas as pd

np.random.seed(42)

def generate_citizen(profile, citizen_id):
    if profile == "trusted":  # 🟢 موثوق
        total = np.random.randint(5, 20)
        valid = np.random.randint(int(0.8*total), total)
        duplicate = np.random.randint(0, 2)
        fake = 0
        recent = np.random.randint(0, 3)
        label = 1

    elif profile == "normal":  # 🟡 عادي
        total = np.random.randint(10, 30)
        valid = np.random.randint(int(0.5*total), int(0.8*total))
        duplicate = np.random.randint(1, 5)
        fake = np.random.randint(0, 2)
        recent = np.random.randint(2, 6)
        label = 1

    else:  # 🔴 spam
        total = np.random.randint(25, 60)
        valid = np.random.randint(0, int(0.4*total))
        duplicate = np.random.randint(5, int(0.5*total))
        fake = np.random.randint(2, 10)
        recent = np.random.randint(10, 25)
        label = 0

    invalid = total - valid
    avg_severity = round(np.random.uniform(1.0, 3.5), 2)

    return {
        "citizen_id": citizen_id,
        "total_reports": total,
        "valid_reports": valid,
        "invalid_reports": invalid,
        "duplicate_reports": duplicate,
        "fake_image_count": fake,
        "avg_severity_reported": avg_severity,
        "reports_last_30_days": recent,
        "trust_label": label
    }

# ===== Generate Dataset =====
data = []
citizen_id = 1

for _ in range(6000):  # 🟢 30%
    data.append(generate_citizen("trusted", citizen_id))
    citizen_id += 1

for _ in range(8000):  # 🟡 40%
    data.append(generate_citizen("normal", citizen_id))
    citizen_id += 1

for _ in range(6000):  # 🔴 30%
    data.append(generate_citizen("spam", citizen_id))
    citizen_id += 1

df = pd.DataFrame(data)

# Feature Engineering
df["valid_ratio"] = df["valid_reports"] / df["total_reports"]
df["duplicate_rate"] = df["duplicate_reports"] / df["total_reports"]
df["fake_image_ratio"] = df["fake_image_count"] / df["total_reports"]

# Save
df.to_csv("citizen_trust_dataset.csv", index=False)

print("✅ Dataset generated:", df.shape)
print(df.head())
```
