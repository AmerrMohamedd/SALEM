```python
import numpy as np
import pandas as pd

np.random.seed(42)

incident_types = ["Electricity", "Water", "Road", "Gas", "Waste"]
area_types = ["Residential", "Commercial", "Industrial"]
teams = ["Team_A", "Team_B", "Team_C"]
priority_map = {"P1": 1.6, "P2": 1.2, "P3": 0.9}

data = []

for i in range(1, 35001):
    incident_type = np.random.choice(incident_types)
    severity_score = round(np.random.uniform(0.2, 1.0), 2)

    priority = np.random.choice(["P1", "P2", "P3"], p=[0.3, 0.4, 0.3])
    assigned_team = np.random.choice(teams)
    area_type = np.random.choice(area_types)

    hour = np.random.randint(0, 24)
    day_of_week = np.random.randint(0, 7)

    area_load = round(np.random.beta(2, 2), 2)
    team_efficiency = round(np.random.beta(4, 2), 2)

    base_time = 2 + severity_score * 6
    priority_factor = priority_map[priority]
    load_factor = 1 + area_load
    efficiency_factor = 1 - (team_efficiency * 0.5)

    night_penalty = 1.2 if hour < 7 or hour > 20 else 1.0
    weekend_penalty = 1.1 if day_of_week in [5, 6] else 1.0

    eta = (
        base_time *
        priority_factor *
        load_factor *
        efficiency_factor *
        night_penalty *
        weekend_penalty
    )

    eta = round(max(0.5, eta), 2)

    data.append({
        "incident_id": i,
        "incident_type": incident_type,
        "severity_score": severity_score,
        "priority_level": priority,
        "assigned_team": assigned_team,
        "area_type": area_type,
        "hour": hour,
        "day_of_week": day_of_week,
        "area_load": area_load,
        "team_efficiency": team_efficiency,
        "estimated_resolution_time": eta
    })

df_eta = pd.DataFrame(data)
df_eta.to_csv("response_time_dataset.csv", index=False)

print("✅ ETA dataset generated:", df_eta.shape)
print(df_eta.head())
```
